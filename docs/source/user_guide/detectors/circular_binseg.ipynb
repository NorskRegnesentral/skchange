# CircularBinarySegmentation

`CircularBinarySegmentation` targets *transient* (a.k.a. *epidemic*) changes: short intervals where the distribution differs from the surrounding baseline. Each detected interval `[start, end)` contributes a pair of changepoints. It searches a seeded grid of outer intervals and, for each, uses an interval scorer of type `transient_score` to test a candidate inner sub-interval against the surrounding data.

## Basic usage

The example below plants a single 10-sample spike in an otherwise stationary Gaussian series and recovers it with [L2TransientScore](../../api_reference/auto_generated/skchange.new_api.interval_scorers.L2TransientScore.rst).

```python
import plotly.io as pio

from skchange.new_api.datasets import generate_piecewise_normal_data
from skchange.new_api.detectors import CircularBinarySegmentation
from skchange.new_api.interval_scorers import L2TransientScore
from skchange.new_api.utils.plotting import plot_detections

pio.renderers.default = "notebook"

X = generate_piecewise_normal_data(
    means=[0, 5, 0, -4, 0],
    lengths=[40, 10, 60, 5, 35],
    seed=2,
)

detector = CircularBinarySegmentation(
    L2TransientScore(),
    min_subinterval_length=3,
    penalty_scale=1.0,
)
detector.fit(X)
anomalies = detector.predict_segment_anomalies(X)

plot_detections(X, segment_anomalies=anomalies).show()
print(anomalies)
```

## Parameters worth knowing

- `transient_score`: an interval scorer of type `transient_score`. Determines how a sub-interval is compared with its surrounding baseline.
- `penalty`, `penalty_scale`, `agg`: same role as in [SeededBinarySegmentation](seeded_binseg.ipynb).
- `min_subinterval_length`: shortest anomalous sub-interval the detector can resolve.
- `max_interval_length`: caps the outer interval size around each candidate sub-interval.
- `growth_factor`: controls how quickly the seeded outer intervals grow.

See the full API reference for [CircularBinarySegmentation](../../api_reference/auto_generated/skchange.new_api.detectors.CircularBinarySegmentation.rst).
