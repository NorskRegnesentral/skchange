# PELT

`PELT` (Pruned Exact Linear Time) minimises the total sum of segment costs plus a per-changepoint penalty. It returns the *globally optimal* segmentation under that objective, with near-linear runtime when the true number of changepoints is proportional to the series length.

Reach for `PELT` when you want an exact segmentation and a cost is a natural fit for the change you're looking for (mean, variance, regression coefficients, ...). It only supports interval scorers of type `cost`.

## Basic usage

The example below detects mean shifts in a piecewise-normal series using [L2Cost](../../api_reference/auto_generated/skchange.new_api.interval_scorers.L2Cost.rst).

```python
import plotly.io as pio

from skchange.new_api.datasets import generate_piecewise_normal_data
from skchange.new_api.detectors import PELT
from skchange.new_api.interval_scorers import L2Cost
from skchange.new_api.utils.plotting import plot_detections

pio.renderers.default = "notebook"

X = generate_piecewise_normal_data(
    means=[0, 10, 0, -3, 5, 1],
    lengths=[30, 5, 15, 50, 60, 40],
    seed=0,
)

detector = PELT(L2Cost(), penalty=10.0)
changepoints = detector.fit_predict(X)

plot_detections(X, changepoints=changepoints).show()
print(changepoints)
```

## Parameters worth knowing

- `cost`: any interval scorer of type `cost`. Determines *what* kind of change is detected.
- `penalty`: the per-changepoint penalty. Larger values produce fewer segments. If left as `None`, a sensible default derived from `cost` is used. See the [Penalties](../concepts/penalties.ipynb) concept page.
- `min_segment_length`: minimum segment length allowed. Useful to avoid very short spurious segments.
- `prune`: enables the pruning step that makes `PELT` fast in practice. Leave at the default `True` unless debugging.
- `step_size`: coarsens the candidate changepoint grid. Larger step sizes are faster but can miss short segments.

See the full API reference for [PELT](../../api_reference/auto_generated/skchange.new_api.detectors.PELT.rst).
