# FPOP
If you have a long series of univariate data and are looking for changes in the mean, `FPOP` is your best option.
It solves the same optimisation problem as [PELT](pelt.ipynb), but is often more than 1000 times faster with our implementations.
It only works for univariate data though, and is so far restricted to the L2 cost function.

## Basic usage

The example below detects mean shifts in a piecewise-normal series. `FPOP` uses the squared-error (L2) cost internally, so no interval scorer is passed.

```python
import plotly.io as pio

from skchange.datasets import generate_piecewise_normal_data
from skchange.detectors import FPOP
from skchange.utils.plotting import plot_detections

pio.renderers.default = "notebook"

X = generate_piecewise_normal_data(
    means=[0, 10, 0, -3, 5, 1],
    lengths=[30, 5, 15, 50, 60, 40],
    seed=0,
)

detector = FPOP()
changepoints = detector.fit_predict(X)

plot_detections(X, changepoints=changepoints).show()
print(changepoints)
```

## Parameters worth knowing

- `penalty`: The per-changepoint penalty. Larger values produce fewer segments. If left as `None`, the BIC penalty `2 * log(n_samples)` is used. See the [Penalties](../concepts/penalties.ipynb) concept page.
- `penalty_scale`: Multiplicative scale applied to the default BIC penalty when `penalty` is `None`. Useful for fine-tuning sensitivity without specifying an absolute penalty value.

See the full API reference for [FPOP](../../api_reference/auto_generated/skchange.detectors.FPOP.rst).

## See also

- [PELT](pelt.ipynb): The same optimal partitioning objective with scalar pruning; supports arbitrary cost-type interval scorers and multivariate input.
- [CROPS](crops.ipynb): Runs the same objective across a range of penalties and returns the full solution path.
